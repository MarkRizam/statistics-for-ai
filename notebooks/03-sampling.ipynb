# Sampling and the central limit theorem

Learning objective: understand and execute the statistical procedure.

For independent exponential samples of mean and variance 1, standardized sample means tend toward a standard normal distribution. KS distance is descriptive here.

```python
import numpy as np
from stats_for_ai import *
import matplotlib.pyplot as plt
from scipy import stats
rng=np.random.default_rng(11)
for size in (5,30,100):
    z=(rng.exponential(size=(2000,size)).mean(1)-1)*np.sqrt(size)
    print(size,stats.kstest(z,"norm").statistic)
    plt.hist(z,bins=35,density=True,histtype="step",label=str(size))
plt.legend(); plt.show()
```
