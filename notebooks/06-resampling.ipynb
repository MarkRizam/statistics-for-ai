# Resampling and paired evaluation

Learning objective: understand and execute the statistical procedure.

Resample independent units with replacement for the percentile bootstrap. A paired sign-flip test requires symmetric independent null differences; the Monte Carlo correction adds one to numerator and denominator.

```python
import numpy as np
from stats_for_ai import *
rng=np.random.default_rng(33); a=rng.normal(size=30); b=a+rng.normal(.3,.5,size=30); print(paired_t(b,a),paired_effect(b,a)); print(sign_flip_test(b-a,1999,11)); print(bootstrap_mean_ci(b-a,seed=11))
```
