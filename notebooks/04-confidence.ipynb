# Confidence intervals

Learning objective: understand and execute the statistical procedure.

A 95% frequentist interval procedure covers a fixed true parameter in about 95% of repeated samples under assumptions; it is not a posterior probability for a single interval.

```python
import numpy as np
from stats_for_ai import *
rng=np.random.default_rng(11); x=rng.normal(size=20); print(mean_ci(x)); print(bootstrap_mean_ci(x,resamples=1000,seed=11))
```
